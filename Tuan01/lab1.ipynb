## Bài 8: Autograd cơ bản 

```python
x = torch.tensor(3.6, requires_grad=True)
y = x * x

y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

```python
grad_manual = 2 * 3.6
print(f"\nĐạo hàm tính thủ công: {grad_manual}")
print(f"Khớp với x.grad: {x.grad.item() == grad_manual}")
```

```python
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2**2 + 2 * x2 + 1

y2.backward()

print("\n YÊU CẦU 2 ")
print("y2 =", y2)
print("x2.grad (PyTorch tính):", x2.grad)

grad_manual_2 = 6 * 3.6 + 2
print(f"Đạo hàm giải tích (6*x + 2): {grad_manual_2}")
```

## Bài 9: Gradient Descent tìm hệ số đa thức

```python
import numpy as np
polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)

learned_w = w.detach().numpy().flatten()
print(f"\nHệ số gốc:       [1.0, 2.0, 3.0]")
print(f"Hệ số học được:  [{learned_w[0]:.4f}, {learned_w[1]:.4f}, {learned_w[2]:.4f}]")
```

## Bài 10: Giải hệ phương trình bằng Autograd

```python
import random
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

a_val, b_val, c_val, d_val = A.item(), B.item(), C.item(), D.item()
print(f"A = {a_val:.4f}")
print(f"B = {b_val:.4f}")
print(f"C = {c_val:.4f}")
print(f"D = {d_val:.4f}")

print("\n KIỂM TRA LẠI CÁC PHƯƠNG TRÌNH ")
print(f"A + B = {a_val + b_val:.4f} (mục tiêu 9)")
print(f"C - D = {c_val - d_val:.4f} (mục tiêu 1)")
print(f"A + C = {a_val + c_val:.4f} (mục tiêu 8)")
print(f"B - D = {b_val - d_val:.4f} (mục tiêu 2)")
print(f"Tổng sai số (sqerr): {sqerr.item():.6e}")
```

## Bài 11: Viết chương trình PyTorch để thực hiện các yêu cầu 

```python
print(" YÊU CẦU 1 ")
X = torch.arange(24) 

x_4x6 = X.reshape(4, 6)
x_2x12 = X.reshape(2, 12)
x_2x3x4 = X.reshape(2, 3, 4)

print("X ban đầu:", X)
print("Reshape (4, 6):\n", x_4x6)
print("Reshape (2, 12):\n", x_2x12)
print("Reshape (2, 3, 4):\n", x_2x3x4)

print("\n YÊU CẦU 2 ")
A = torch.rand(3, 4) + 0.1
B = torch.rand(3, 4) + 0.1

print("A + B:\n", A + B)
print("A * B (Nhân từng phần tử):\n", A * B)
print("A / B:\n", A / B)


print("Mean của A theo cột (dim=0):", torch.mean(A, dim=0))
print("Std của A theo cột (dim=0):", torch.std(A, dim=0))

print("\n YÊU CẦU 3 ")
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3

y.backward()

grad_manual = 4 * 2.0 + 5.0

print(f"y = {y.item()}")
print(f"Gradient tính bởi Autograd (x.grad): {x.grad.item()}")
print(f"Gradient tính thủ công (4x + 5):     {grad_manual}")
print(f"Kết quả trùng khớp: {x.grad.item() == grad_manual}")

print("\n YÊU CẦU 4 ")
polynomial = np.poly1d([2, -3, 5])
N = 30

X_train = np.random.randn(N, 1) * 5
Y_train = polynomial(X_train)

XX = np.hstack([
    X_train * X_train,
    X_train,
    np.ones_like(X_train)
])

x_tensor = torch.tensor(XX, dtype=torch.float32)
y_tensor = torch.tensor(Y_train, dtype=torch.float32)

w = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.NAdam([w], lr=0.05)

for epoch in range(2000):
    optimizer.zero_grad()
    y_pred = x_tensor @ w
    mse = torch.mean(torch.square(y_tensor - y_pred))
    mse.backward()
    optimizer.step()

learned_coeffs = w.detach().numpy().flatten()
print("Hệ số gốc cần tìm: [2.0, -3.0, 5.0]")
print(f"Hệ số học được:    [{learned_coeffs[0]:.4f}, {learned_coeffs[1]:.4f}, {learned_coeffs[2]:.4f}]")
print(f"Hàm mất mát MSE cuối cùng: {mse.item():.6e}")
```
